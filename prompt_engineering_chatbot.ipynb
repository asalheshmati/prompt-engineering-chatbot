## Set Up OpenAI API

Configure the OpenAI API key and import necessary libraries to interact with the OpenAI models.

```python
from dotenv import load_dotenv
from openai import OpenAI
import os

# Load environment variables from .env file
load_dotenv()

# Initialize OpenAI client
client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'))

if not os.getenv('OPENAI_API_KEY'):
    print("Please set the OPENAI_API_KEY environment variable.")
else:
    print("OpenAI API configured successfully.")
```

## Implement Sentiment Analysis with Few-Shot Prompting

Create a prompt for classifying customer messages as positive, negative, or neutral using few-shot examples.

```python
# Define the sentiment analysis prompt with few-shot examples
sentiment_prompt_template = """
Classify the sentiment of the following customer message as positive, negative, or neutral.

Examples:
Message: "I love this product! It works perfectly."
Sentiment: positive

Message: "This is terrible. It broke after one use."
Sentiment: negative

Message: "The product arrived on time."
Sentiment: neutral

Now, classify this message:
{message}
Sentiment:
"""

def analyze_sentiment(message):
    prompt = sentiment_prompt_template.format(message=message)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=10,
        temperature=0
    )
    return response.choices[0].message.content.strip().lower()

# Test the prompt
test_message = "I'm very happy with the service."
print(f"Message: {test_message}")
print(f"Sentiment: {analyze_sentiment(test_message)}")
```

## Test Sentiment Analysis Prompt Reliability

Run the sentiment analysis prompt multiple times (5, 10, 15 iterations) and analyze patterns for consistency.

```python
import collections

# Test the sentiment analysis prompt 15 times
test_message = "This product is okay, nothing special."
results = []
for i in range(15):
    result = analyze_sentiment(test_message)
    results.append(result)
    print(f"Iteration {i+1}: {result}")

# Analyze patterns
counter = collections.Counter(results)
print("\nSentiment distribution:")
for sentiment, count in counter.items():
    print(f"{sentiment}: {count} times")

print(f"\nConsistency: {max(counter.values()) / len(results) * 100:.1f}% of responses were the same.")
```

## Apply Chain-of-Thought Prompting to Sentiment Analysis

Enhance the sentiment analysis prompt with Chain-of-Thought reasoning to improve accuracy.

```python
# Define the CoT sentiment analysis prompt
cot_sentiment_prompt_template = """
Classify the sentiment of the following customer message as positive, negative, or neutral. Use Chain-of-Thought reasoning.

Examples:
Message: "I love this product! It works perfectly."
Reasoning: The message contains positive words like "love" and "perfectly", expressing satisfaction. Overall tone is enthusiastic.
Sentiment: positive

Message: "This is terrible. It broke after one use."
Reasoning: Words like "terrible" and "broke" indicate dissatisfaction and failure. Tone is negative.
Sentiment: negative

Message: "The product arrived on time."
Reasoning: "On time" is factual, no strong positive or negative emotions expressed.
Sentiment: neutral

Now, classify this message:
{message}

Reasoning:
Sentiment:
"""

def analyze_sentiment_cot(message):
    prompt = cot_sentiment_prompt_template.format(message=message)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=100,
        temperature=0
    )
    return response.choices[0].message.content.strip()

# Test the CoT prompt
print("Chain-of-Thought Sentiment Analysis:")
print(f"Message: {test_message}")
print(analyze_sentiment_cot(test_message))
```

## Implement Product Description Generation

Develop a prompt to generate compelling product descriptions from basic product information.

```python
# Define the product description generation prompt
product_desc_prompt_template = """
Generate a compelling product description based on the following basic product information. Make it engaging, highlight key benefits, and use persuasive language suitable for an e-commerce website.

Product Information:
{product_info}

Product Description:
"""

def generate_product_description(product_info):
    prompt = product_desc_prompt_template.format(product_info=product_info)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=200,
        temperature=0.7
    )
    return response.choices[0].message.content.strip()

# Test the prompt
test_product = "Name: Wireless Bluetooth Headphones\nFeatures: Noise cancellation, 30-hour battery, comfortable fit\nPrice: $99"
print("Product Description Generation:")
print(generate_product_description(test_product))
```

## Tune Product Description Generation with Constraints

Add structure and constraints to the product description prompt for more reliable outputs.

```python
# Tuned product description prompt with constraints
tuned_product_desc_prompt_template = """
Generate a compelling product description based on the following basic product information. Follow these constraints:
- Length: 100-150 words
- Structure: Start with an engaging hook, list key features, highlight benefits, end with a call to action
- Tone: Persuasive and professional
- Include: Price if provided

Product Information:
{product_info}

Product Description:
"""

def generate_product_description_tuned(product_info):
    prompt = tuned_product_desc_prompt_template.format(product_info=product_info)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=250,
        temperature=0.7
    )
    return response.choices[0].message.content.strip()

# Test the tuned prompt
print("Tuned Product Description Generation:")
print(generate_product_description_tuned(test_product))
```

## Test Product Description Generation Iteratively

Iterate the product description prompt multiple times and evaluate variations for quality.

```python
# Test the tuned product description prompt 5 times
print("Iterative Testing of Product Description Generation:")
for i in range(5):
    desc = generate_product_description_tuned(test_product)
    word_count = len(desc.split())
    print(f"\nIteration {i+1} (Word count: {word_count}):")
    print(desc)
```

## Implement Data Extraction from Feedback

Create a prompt to extract structured information from unstructured customer feedback.

```python
# Define the data extraction prompt
data_extraction_prompt_template = """
Extract structured information from the following customer feedback. Identify:
- Product/Service mentioned
- Main issue or complaint
- Sentiment (positive/negative/neutral)
- Any suggestions or recommendations

Feedback: {feedback}

Extracted Information:
- Product/Service: 
- Issue: 
- Sentiment: 
- Suggestion: 
"""

def extract_feedback_data(feedback):
    prompt = data_extraction_prompt_template.format(feedback=feedback)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=150,
        temperature=0
    )
    return response.choices[0].message.content.strip()

# Test the prompt
test_feedback = "I bought the wireless headphones last week, but they keep disconnecting randomly. It's really frustrating. Please fix this issue or offer a refund."
print("Data Extraction from Feedback:")
print(extract_feedback_data(test_feedback))
```

## Add Structure to Data Extraction Prompts

Incorporate formatting and constraints into the data extraction prompt for better reliability.

```python
# Structured data extraction prompt
structured_data_extraction_prompt_template = """
Extract structured information from the following customer feedback. Output in JSON format with the following keys:
- "product": Product or service mentioned
- "issue": Main issue or complaint (if any)
- "sentiment": positive, negative, or neutral
- "suggestion": Any suggestions or recommendations (if none, use "None")

Feedback: {feedback}

JSON Output:
"""

def extract_feedback_data_structured(feedback):
    prompt = structured_data_extraction_prompt_template.format(feedback=feedback)
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=200,
        temperature=0
    )
    return response.choices[0].message.content.strip()

# Test the structured prompt
print("Structured Data Extraction from Feedback:")
result = extract_feedback_data_structured(test_feedback)
print(result)

# Optionally parse JSON
import json
try:
    parsed = json.loads(result)
    print("Parsed JSON:", parsed)
except:
    print("Could not parse as JSON")
```

## Diagnose and Iterate on Data Extraction Prompts

Run the data extraction prompt iteratively (5, 10, 15 times) to identify failures and refine it.

```python
# Test the structured data extraction prompt 15 times
print("Iterative Testing of Data Extraction:")
success_count = 0
for i in range(15):
    result = extract_feedback_data_structured(test_feedback)
    try:
        parsed = json.loads(result)
        success = True
        success_count += 1
    except:
        success = False
    print(f"Iteration {i+1}: {'Valid JSON' if success else 'Invalid JSON'}")
    if not success:
        print(f"Output: {result}")

print(f"\nJSON Parsing Success Rate: {success_count}/15 ({success_count/15*100:.1f}%)")
```
